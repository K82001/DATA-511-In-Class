# Exercise 3 (5 points)

Write a Python function to simulate tossing a coin 10,000 times. Use 0 for tails and 1 for heads.

```python
import numpy as np

#simulate a series of coin flips (0= tails, 1= heads)

n_sim = 10000
results = [np.random.choice([0,1], p = [0.5, 0.5]) for _ in range(n_sim)]

# calculate the proportion of heads 
prop_heads = np.mean(results)
print(f"Proportion of heads: {prop_heads}")
```

Using Python, simulate 1000 regressions and store coefficients of each run.

```python
import pandas as pd
import statsmodels.formula.api as smf
import matplotlib.pyplot as plt

betas = []
for i in range(1000):
    #Simulate some data
    np.random.seed(i)
    x = np.random.uniform(0.6, 0.95, size = 120)
    y = 50 + 20*x + np.random.normal(0, 2, size = 120)


    df = pd.DataFrame({
        #'x': np.random.normal(size = 100)
        #'y': np.random.normal(size = 100)
        'x' : x,
        'y' : y
    })

    model = smf.ols('y ~ x', data = df).fit()
    betas.append(model.params['x'])

plt.hist(betas, bins=30, color="steelblue")
plt.xlabel("Beta Estimates")
plt.ylabel("Frequency")
plt.title("Distribution of Beta Estimates from 1000 Simulations")
plt.vlines(x=20, ymin=0, ymax=max(plt.hist(betas, bins=30) [0]), colors = 'r', linestyles='dashed', label='True Coefficient', linewidth=2)
plt.legend()
plt.grid()
plt.show();
```
