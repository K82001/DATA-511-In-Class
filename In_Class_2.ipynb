# Exercise 2 (4 points)

Briefly describe the difference between stochastic and non-stochastic prescriptive methods.

Non-stochastic prescriptive methods do not directly incorporate randomness and focus on deterministic processes  (e.g - mathematical programming).  Stochastic methods explicitly account for random variations such as Monte Carlo simulations or probabilistic modeling.

# Exercise 4 (10 points)

Consider the `train.csv` data file. This data file contains information related to music tracks. The goal is to build a machine learning model to predict `BeatsPerMinute`. Explore the data.

```python
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv("train.csv")
df.head()
```

```python
plt.hist(df["BeatsPerMinute"], bins=20, edgecolor="black", color="steelblue")
plt.xlabel("Beats Per Minute")
plt.ylabel("Frequency")
plt.title("Distribution of Beats Per Minute")
plt.grid()
plt.show()
```

The "Beats per Minute" distrubution looks symetrical and normally distributed.

```python
plt.plot(df["Energy"], df["BeatsPerMinute"], "o", color="steelblue")
plt.xlabel("Energy")
plt.ylabel("BeatsPerMinute")
plt.title("Beats Per Minute vs Energy")
plt.show()
```
